```python
import tkinter as tk
from tkinter import messagebox

def calculate_bmi():
    name = name_entry.get()
    try:
        weight = float(weight_entry.get())
        height = float(height_entry.get())

        bmi = weight / (height ** 2)
        bmi_result = f"{name}, your BMI is {bmi:.2f}\n"

        if bmi < 18.5:
            bmi_result += "You are underweight."
        elif bmi < 25:
            bmi_result += "You are healthy."
        elif bmi < 30:
            bmi_result += "You are overweight."
        else:
            bmi_result += "You are obese."

        messagebox.showinfo("BMI Result", bmi_result)
    except ValueError:
        messagebox.showerror("Invalid Input", "Please enter valid numbers for weight and height.")

# Set up main window
root = tk.Tk()
root.title("BMI Calculator")
root.geometry("400x300")
root.configure(bg="lightblue")

# Labels and entry fields
tk.Label(root, text="Enter your name:", bg="lightblue", font=("Arial", 12)).pack(pady=5)
name_entry = tk.Entry(root, font=("Arial", 12))
name_entry.pack()

tk.Label(root, text="Enter your weight (kg):", bg="lightblue", font=("Arial", 12)).pack(pady=5)
weight_entry = tk.Entry(root, font=("Arial", 12))
weight_entry.pack()

tk.Label(root, text="Enter your height (m):", bg="lightblue", font=("Arial", 12)).pack(pady=5)
height_entry = tk.Entry(root, font=("Arial", 12))
height_entry.pack()

tk.Button(root, text="Calculate BMI", command=calculate_bmi, font=("Arial", 12), bg="#4CAF50", fg="white").pack(pady=20)

root.mainloop()
```
